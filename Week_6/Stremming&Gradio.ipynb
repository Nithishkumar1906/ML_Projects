````python
def stream_markdown(prompt):
    """
    Stream Markdown content from Gemini in real-time.

    Args:
        prompt (str): The input prompt for the Gemini model.

    Returns:
        None
    """
    model="gemini-2.5-flash-lite"
    stream = Gemini.generate_content(
        contents=[prompt],
        stream = True
    )

    buffer = ""
    display_handle = display(Markdown(""), display_id=True)

    for chunk in stream:
        buffer += chunk.text
        buffer = buffer.replace("```", "")
        display_handle.update(Markdown(buffer))

prompt = "Generate a short brochure in markdown for a LLM models."
stream_markdown(prompt)
````

````python
GEMINI_MODEL = "gemini-2.5-flash-lite"   
OLLAMA_MODEL = "llama3.2"                  
OLLAMA_CHAT_URL = "http://localhost:11434/api/generate"


ARENA_PREAMBLE = """\
You are one of two AI debaters. Your goal is to PERSUADE the audience that your view is stronger than your opponent's.
Rules:
1. Start by briefly stating your position.
2. Identify specific weaknesses, gaps, or biases in your opponent's last answer (if given).
3. Present counterarguments or evidence supporting your own position.
4. Maintain a professional but competitive tone.
5. End with a strong closing point that differentiates your stance from theirs.

Do NOT simply repeat their points in different words — directly engage and challenge them.
"""

def build_turn_prompt(role_name, user_task, opponent_last=None):
    parts = [
        f"Role: {role_name}",
        ARENA_PREAMBLE,
        f"User task:\n{user_task}",
    ]
    if opponent_last:
        parts.append(f"Opponent's last answer:\n{opponent_last}")
    return "\n\n".join(parts)


def ask_gemini(prompt: str) -> str:
    model = genai.GenerativeModel(GEMINI_MODEL)
    resp = model.generate_content(prompt)
    return (resp.text or "").strip()


def ask_ollama(prompt: str) -> str:
    payload = {
        "model": OLLAMA_MODEL,
        "prompt": prompt,
        "stream": False
    }
    r = requests.post(OLLAMA_CHAT_URL, json=payload)
    r.raise_for_status()
    data = r.json()
    return data.get("response", "").strip()
````

````python
def adversarial_convo(user_task: str, rounds: int = 3, start_with="gemini"):
    """
    Runs a small debate where Gemini and Ollama alternate producing improved answers.
    Returns the full transcript and the final answer.
    """
    transcript = []
    last_answer = None
    speaker = start_with.lower()

    for i in range(1, rounds + 1):
        if speaker == "gemini":
            prompt = build_turn_prompt("Gemini", user_task, opponent_last=last_answer)
            answer = ask_gemini(prompt)
            transcript.append(("Gemini", answer))
            last_answer = answer
            speaker = "ollama"
        else:
            prompt = build_turn_prompt("Ollama", user_task, opponent_last=last_answer)
            answer = ask_ollama(prompt)
            transcript.append(("Ollama", answer))
            last_answer = answer
            speaker = "gemini"

    return transcript, last_answer
````

````python
task = """Explain the trade off between AI is Boon or Bone.
Provide when to prefer each, and common pitfalls. Keep it under 180 words."""
transcript, final_answer = adversarial_convo(task, rounds=4, start_with="gemini")

for who, msg in transcript:
    print(f"\n=== {who} ===\n{msg}\n")

print("\n=== FINAL ANSWER ===\n", final_answer)
````

````python
GEMINI_MODEL = "gemini-2.5-flash-lite"   
OLLAMA_MODEL = "llama3.2"                  

def ask_gemini_1(prompt: str, model: str) -> str:
    """
    Send a prompt to Google's Gemini AI model and return the response.
    
    Args:
        prompt (str): The text prompt to send to the Gemini model
        model (str): The specific Gemini model to use (e.g., 'gemini-pro')
    
    Returns:
        str: The model's response text, stripped of leading/trailing whitespace
        
    Raises:
        Exception: If the API call fails or returns an invalid response
    """
    gem = genai.GenerativeModel(model)
    resp = gem.generate_content(prompt)
    return (resp.text or "").strip()

def ask_ollama_1(prompt: str, model: str) -> str:
    """
    Send a prompt to a local Ollama model and return the response.
    
    Args:
        prompt (str): The text prompt to send to the Ollama model
        model (str): The specific Ollama model to use (e.g., 'llama2')
    
    Returns:
        str: The model's response text, stripped of leading/trailing whitespace
        
    Raises:
        requests.HTTPError: If the HTTP request to Ollama fails
        requests.ConnectionError: If unable to connect to Ollama server
    """
    payload = {
        "model": model,
        "prompt": prompt,
        "stream": False
    }
    r = requests.post("http://localhost:11434/api/generate", json=payload)
    r.raise_for_status()
    data = r.json()
    return data.get("response", "").strip()
````

````python
def build_turn_prompt(speaker_name, task, opponent_last=None):
    """
    Construct a debate prompt that assigns opposing positions to create adversarial dialogue.
    
    Gemini is assigned the PRO-AI position (AI as primarily beneficial)
    Ollama is assigned the ANTI-AI position (AI as primarily risky)
    
    Args:
        speaker_name (str): Name of the AI model speaking ('Gemini' or 'Ollama')
        task (str): The main debate topic or question
        opponent_last (str, optional): The previous response from the opposing AI
    
    Returns:
        str: A formatted prompt that encourages the AI to take a specific stance
             and engage in adversarial debate
    """
    if speaker_name.lower() == "gemini":
        position = "Gemini-AI (argue that AI is primarily a BOON)"
        counter_instruction = "Challenge and refute your opponent's arguments while strengthening your PRO-AI stance."
    else:  # Ollama
        position = "Ollama-AI (argue that AI is primarily a RISK)" 
        counter_instruction = "Challenge and refute your opponent's arguments while strengthening your ANTI-AI stance."
    
    if opponent_last:
        return f"""You are {speaker_name} taking the {position} position in this debate.

Your opponent just said:
{opponent_last}

{counter_instruction}

Debate topic: {task}

Respond with a strong, persuasive argument for your assigned position. Be direct, confident, and challenge your opponent's points."""
    else:
        return f"""You are {speaker_name} taking the {position} position in this debate.

Debate topic: {task}

Present a strong, persuasive opening argument for your assigned position. Be direct, confident, and make compelling points."""
````

````python
def adversarial_convo_stream(user_task, rounds=3, start_with="gemini",
                             gem_model=GEMINI_MODEL, oll_model=OLLAMA_MODEL):
    """
    Generate an adversarial conversation between Gemini and Ollama models.
    
    Creates a streaming debate where the two AI models take opposing positions
    and respond to each other in alternating turns. Yields the complete transcript
    after each turn for real-time display.
    
    Args:
        user_task (str): The debate topic or question to discuss
        rounds (int, optional): Total number of turns in the debate. Defaults to 3.
        start_with (str, optional): Which AI should start ('gemini' or 'ollama'). 
                                  Defaults to 'gemini'.
        gem_model (str, optional): Gemini model name. Defaults to GEMINI_MODEL.
        oll_model (str, optional): Ollama model name. Defaults to OLLAMA_MODEL.
    
    Yields:
        list[tuple[str, str]]: Complete conversation transcript as list of 
                              (speaker_name, response) tuples after each turn
    
    Example:
        >>> for transcript in adversarial_convo_stream("Is AI safe?", rounds=2):
        ...     print(f"Current transcript has {len(transcript)} exchanges")
        Current transcript has 1 exchanges
        Current transcript has 2 exchanges
    """
    transcript = []
    last_answer = None
    speaker = start_with.lower()
    
    for _ in range(1, rounds + 1):
        if speaker == "gemini":
            prompt = build_turn_prompt("Gemini", user_task, opponent_last=last_answer)
            answer = ask_gemini_1(prompt, model=gem_model)  # Fixed function name
            transcript.append(("Gemini", answer))
            yield transcript[:]  # always yield a list of tuples
            last_answer = answer
            speaker = "ollama"
        else:
            prompt = build_turn_prompt("Ollama", user_task, opponent_last=last_answer)
            answer = ask_ollama_1(prompt, model=oll_model)  # Fixed function name
            transcript.append(("Ollama", answer))
            yield transcript[:]
            last_answer = answer
            speaker = "gemini"
````

````python
with gr.Blocks(theme=gr.themes.Soft(), fill_height=True) as demo:
    gr.Markdown("# 🥊 Gemini vs Ollama — Live Debate")
    gr.Markdown(
        "Enter a debate topic, choose models, and watch the round-by-round exchange."
    )
    with gr.Row():
        topic = gr.Textbox(
            label="Debate Topic / Task",
            placeholder="e.g., Is AI regulation a boon or a barrier?",
            lines=3,
            value="Explain whether AI is more of a boon or a risk. Provide when to prefer each, and common pitfalls. Limit to ~150 words."
        )
    with gr.Row():
        gem_model = gr.Textbox(label="Gemini model", value=GEMINI_MODEL)
        oll_model = gr.Textbox(label="Ollama model", value=OLLAMA_MODEL)
    with gr.Row():
        rounds = gr.Slider(1, 8, value=4, step=1, label="Rounds (total turns)")
        start = gr.Dropdown(
            ["gemini", "ollama"],
            value="gemini",
            label="Start with"
        )
    chat = gr.Chatbot(label="Debate Transcript", height=420, type="tuples")
    run_btn = gr.Button("Start Debate 🔥", variant="primary")
    
    def run_debate(user_task, r, starter, g_model, o_model):
        """
        Execute a live debate between AI models and stream results to Gradio interface.
        
        This function orchestrates the debate by calling adversarial_convo_stream
        and formatting the responses for display in the Gradio chatbot component.
        
        Args:
            user_task (str): The debate topic or question
            r (float): Number of rounds (converted to int internally)
            starter (str): Which AI model should start the debate ('gemini' or 'ollama')
            g_model (str): The Gemini model name to use
            o_model (str): The Ollama model name to use
        
        Yields:
            list[tuple[str, str]]: Formatted messages for Gradio chatbot display,
                                  where each tuple is (formatted_speaker, message)
        
        Note:
            - First yields an empty list to clear the chat interface
            - Then yields progressively longer lists as the debate continues
            - Speaker names are formatted with markdown bold for display
        """
        yield []
        for partial in adversarial_convo_stream(user_task, int(r), starter, g_model, o_model):
            ui_msgs = []
            for who, msg in partial:
                ui_msgs.append((f"**{who}**", msg))
            yield ui_msgs
    
    run_btn.click(
        fn=run_debate,
        inputs=[topic, rounds, start, gem_model, oll_model],
        outputs=chat
    )

demo.queue().launch()
````
