**Practicing Transfromers and HuggingFace**

```python
import torch
from transformers import pipeline
from diffusers import DiffusionPipeline
from datasets import load_dataset
import soundfile as sf
from IPython.display import Audio
```

```python
#Sentiment Analysis

classifier = pipeline("sentiment-analysis")
result = classifier("I've been waiting for a HuggingFace course my whole life.")
print(result)
```

```python
#Named Entity Recognition

ner = pipeline("ner", grouped_entities=True)
ner("My name is Nithish and I work at Hugging Face in Bangalore.")
```

```python
#Question & Answering Bot
QA = pipeline("question-answering")
QA(
    question="What is my team contribution??",
    context="""Nithish is a professional working for Hugging Face, an influential American company at the forefront of the artificial intelligence community.
    He is based in Bangalore, a city often called the "Silicon Valley of India" due to its high concentration of tech companies and startups.
    Hugging Face is widely recognized for its mission to democratize good machine learning. The company is best known for its open-source library, Transformers, which provides extensive tools and resources for building, training, and deploying state-of-the-art machine learning models, particularly in the field of Natural Language Processing (NLP).
    by working at their Bangalore office, Nithish is part of a global team contributing to a collaborative platform that hosts thousands of pre-trained models and datasets, empowering developers and researchers worldwide. His role could potentially involve areas such as machine learning engineering, developer advocacy, or research,
    contributing to the vibrant open-source AI ecosystem that Hugging Face champions."""
)
```

```python
#Text Summarize
summarize = pipeline("summarization")
summarize("""Nithish is professionally associated with Hugging Face, a key player in the global artificial intelligence landscape, and carries out his responsibilities from their office in Bangalore, India. His role places him within one of the world's most dynamic technology ecosystems, contributing to a company whose impact is felt across the entire machine learning community. Hugging Face's primary contribution is its collaborative, open-source platform, centered around the Hugging Face Hub, which serves as a central repository for models, datasets, and applications. This ecosystem is designed to foster transparency and collaboration, enabling researchers and developers to share and build upon each other's work effortlessly. As a part of this organization, Nithish is involved in a mission to advance AI by making powerful tools and state-of-the-art models, such as those found in the renowned Transformers library, universally accessible. His work in Bangalore supports the engineering, research, and community efforts that empower countless individuals and organizations to build sophisticated AI applications in fields ranging from natural language understanding to computer vision and beyond..""")
```

```python
#Translations
translate = pipeline("translation",model="gopi30/english-to-tamil-stage4",src_lang="en",tgt_lang="ta")
tamil_translate = translate("Hugging Face is a French company based in New-York.")
print(tamil_translate)
```

```python
#classification
classification = pipeline("zero-shot-classification")
classification(
    "Hugging Face Transformers library is amazing",
    candidate_labels=["education", "politics", "Technology"],
)
```
